```python
# Import Pandas Libraries 

import pandas as pd 
```

```python
# Load the dataset (Superstore) for Analysis

df = pd.read_csv("Sample - Superstore.csv", encoding="Latin1")
```

```python
# Analysis 

print("Total Sales:", df["Sales"].sum())
print("Average Sales:", df["Sales"].mean())
print("Maximum Sales:", df["Sales"].max())
print("Minimum Sales:", df["Sales"].min())
print("Total Records:", len(df))
```

```python
# Business Insights

# 1- Technology category generated the highest sales compared to
#  Furniture and Office Supplies.

# 2- The dataset contains a large number of sales records, indicating 
# a wide range of customer transactions.

# 3- Sales values vary significantly, as the maximum sales amount is much
#  higher than the minimum sales amount.

# 4- The average sales value provides an overview of the typical order 
# amount placed by customers.

# 5- Products are sold across multiple regions and categories, showing that
#  the business serves a diverse customer base.
```
