```python
# Import Pandas Library

import pandas as pd 
```

```python
# Load the dataset (Superstore)

df = pd.read_csv("Sample - Superstore.csv", encoding="Latin1")
```

```python
# Total SUM of Sales 

total_sales = df["Sales"].sum()
print("Total Sales:",total_sales)
```

```python
# Sum of Profit

total_profit = df["Profit"].sum()
print("Total Profit:", total_profit)
```

```python
# Sum of Quantity

total_quantity = df["Quantity"].sum()
print("Total Quantity:", total_quantity)
```

```python
# Average Sales 

avg_sales = df["Sales"].mean()
print("Averge Sales:", avg_sales)
```

```python
# Averge Profit 

avg_profit = df["Profit"].mean()
print("Average Profit:", avg_profit)
```

```python
# Average Quantity

avg_quantity = df["Quantity"].mean()
print("Average Quantity:", avg_quantity)
```

```python
# maximum Sales 

max_sales = df["Sales"].max()
print("Maximum Sales:", max_sales)
```

```python
# MAximum Profit 

max_profit = df["Profit"].max()
print("Maximum Profit:", max_profit)
```

```python
# Maximum Quantity 

max_quantity = df["Quantity"].max()
print("Maximum Quantity:", max_quantity)
```

```python
# Minimum Sales

min_sales = df["Sales"].min()
print("Minimum Sales:", min_sales)
```

```python
# Minimum Profit 

min_profit = df["Profit"].min()
print("Minimum Profit:", min_profit)
```

```python
# Minimum Quantity 

min_quantity  = df["Quantity"].min()
print("Minimum Quantity:", min_quantity)
```

```python
# count sales

count_sales = df["Sales"].count()
print("Total No of Sales:", count_sales)
```

```python
# Business Insights:

# 1. The dataset contains a large number of sales records.
# 2. The total sales indicate strong overall business performance.
# 3. The average profit shows the average earnings per order.
# 4. The highest sales value is much greater than the minimum sales value, indicating variation in order values.
# 5. The total quantity sold reflects a significant number of products purchased by customers.
```
