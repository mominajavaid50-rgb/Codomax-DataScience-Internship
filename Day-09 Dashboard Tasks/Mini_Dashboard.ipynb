```python
# Import Matplotlib and Pandas Libraries 

import matplotlib.pyplot as plt
import pandas as pd 
```

```python
# Load the Dataset (Superstore)

df = pd.read_csv("Sample - Superstore.csv", encoding="Latin1")
```

```python
# Sales Analysis 

total_sales = df["Sales"].sum()
print("Total Sales:",total_sales)
avg_sales = df["Sales"].mean()
print("Averge Sales:", avg_sales)
max_sales = df["Sales"].max()
print("Maximum Sales:", max_sales)
min_sales = df["Sales"].min()
print("Minimum Sales:", min_sales)
```

```python
# Profit Analysis 

total_profit = df["Profit"].sum()
print("Total Profit:", total_profit)
avg_profit = df["Profit"].mean()
print("Average Profit:", avg_profit)
max_profit = df["Profit"].max()
print("Maximum Profit:", max_profit)
min_profit = df["Profit"].min()
print("Minimum Profit:", min_profit)
```

```python
# Quantity Analysis 

total_quantity = df["Quantity"].sum()
print("Total Quantity:", total_quantity)
avg_quantity = df["Quantity"].mean()
print("Average Quantity:", avg_quantity)
max_quantity = df["Quantity"].max()
print("Maximum Quantity:", max_quantity)
min_quantity  = df["Quantity"].min()
print("Minimum Quantity:", min_quantity)
```

```python
# Bar Chart Total Sales by Category 

sales_category = df.groupby("Category")["Sales"].sum()
plt.figure(figsize=(6,5))
plt.bar(sales_category.index,
        sales_category.values,
        color=["skyblue","lightyellow","pink"])
plt.title("Total Sales by Category")
plt.xlabel("Category")
plt.ylabel("Sales")
plt.show()
```

```python
# Line Chart Profit by Region

profit_region = df.groupby("Region")["Profit"].sum()
plt.figure(figsize=(6,5))
plt.plot(profit_region.index,
         profit_region.values,
         marker="o",
         color="blue")
plt.title("Profit BY Region")
plt.xlabel("Region")
plt.ylabel("Profit")
plt.show()
```

```python
# Pie Chart (all catgories sales)

sales_category = df.groupby("Category")["Sales"].sum()
plt.figure(figsize=(6,5))
plt.pie(sales_category.values,
        labels=sales_category.index,
        autopct="%1.1f%%",
       colors = ["gold", "coral", "mediumpurple"])
plt.title("Profit by Region")
plt.show()
```
