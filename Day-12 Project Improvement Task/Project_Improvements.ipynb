```python
# ======================================
# Import Pandas & Matplotlib Libraries
# ======================================

import pandas as pd 
import matplotlib.pyplot as plt
```

```python
# ===============================
# Load the DataSet (Superstore)
# ===============================

df = pd.read_csv("Sample - Superstore.csv", encoding="Latin1")
```

```python
# ===================
# Data Cleaning 
# ===================

df.isnull().sum()
df.duplicated().sum()
df.drop_duplicates()
```

```python
# ==================
# Data Analysis
# ==================

total_sales = df["Sales"].sum()
print("Total Sales:",total_sales)
min_sales = df["Sales"].min()
print("Minimum Sales:", min_sales)
max_profit = df["Profit"].max()
print("Maximum Profit:", max_profit)
avg_sales = df["Sales"].mean()
print("Average Sales:",avg_sales)
print("Total Records:", len(df))
```

```python
# ================================
# Data Visualization (Graphs)
# ================================
# Line Chart Region-wise Profit
# ================================

region_profit = df.groupby("Region")["Profit"].sum()

plt.figure(figsize=(4,3))

plt.plot(
    region_profit.index,
    region_profit.values,
    marker="o",
    markersize=8,
    linewidth=3,
    color="#4E79A7"
)

plt.title("Profit by Region", fontsize=15, fontweight="bold")
plt.xlabel("Region", fontsize=12)
plt.ylabel("Profit", fontsize=12)

plt.grid(True, linestyle="--", alpha=0.5)

plt.show()
```

```python
# ================================
# Line Chart Region-Wise Profit
# ================================

# Category-wise Sales
category_sales = df.groupby("Category")["Sales"].sum()

plt.figure(figsize=(4,3))

plt.bar(
    category_sales.index,
    category_sales.values,
    color=["#DB64C1", "#28AEAC", "#A165C6"],
    edgecolor="black",
    linewidth=1.2
)

plt.title("Total Sales by Category", fontsize=15, fontweight="bold")
plt.xlabel("Category", fontsize=12)
plt.ylabel("Sales", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.show()
```
