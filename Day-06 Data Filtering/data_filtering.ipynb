```python
# Import Pandas Library 

import pandas as pd
```

```python
# Load the Dataset (Superstore)

df = pd.read_csv("Sample - Superstore.csv", encoding="Latin1")
```

```python
# Filter Rows (sales greater than 10000)

filter_sales = df[df["Sales"]>10000]
filter_sales
```

```python
# Filter profit rows greater than 6000

profit_rows = df[df["Profit"]>6000]
profit_rows
```

```python
# Filter rows where Quantity greaer than 13

quantity_rows = df[df["Quantity"]>13]
quantity_rows
```

```python
# Where Ship mode is second class

ship_mode = df[df["Ship Mode"]=="Second Class"]
ship_mode
```

```python
# where region is west and sales are greater than 5000

west_sales = df[(df["Region"]=="West") & (df["Sales"] > 5000)]
west_sales
```
